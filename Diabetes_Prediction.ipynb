```python

# Replace zero values with the median
dataframe['Pregnancies'] = dataframe['Pregnancies'].replace(
    0, dataframe['Pregnancies'].median()
)

dataframe['Insulin'] = dataframe['Insulin'].replace(
    0, dataframe['Insulin'].median()
)

dataframe['BloodPressure'] = dataframe['BloodPressure'].replace(
    0, dataframe['BloodPressure'].median()
)

dataframe['SkinThickness'] = dataframe['SkinThickness'].replace(
    0, dataframe['SkinThickness'].median()
)

# Replace zero BMI values with the mean
dataframe['BMI'] = dataframe['BMI'].replace(
    0, dataframe['BMI'].mean()
)

dataframe['DiabetesPedigreeFunction'] = dataframe[
    'DiabetesPedigreeFunction'
].replace(
    0, dataframe['DiabetesPedigreeFunction'].mean()
)

dataframe['Age'] = dataframe['Age'].replace(
    0, dataframe['Age'].median()
)
```

**Outlier Detection and Normaliztion**

```python
dataframe.shape
```

```python
dataframe.columns
```

```python
# Remove the incorrectly created column, if it exists
if 'Pregnacies' in dataframe.columns:
    dataframe = dataframe.drop(columns=['Pregnacies'])

print(dataframe.shape)
print(dataframe.columns.tolist())
```

```python
X = dataframe.drop(columns=['Outcome'])
y = dataframe['Outcome']
```

```python
dataframe.columns
```

```python
fig, ax = plt.subplots(figsize = (15, 15))
sns.boxplot(data = X, ax=ax)
plt.savefig('boxPlot.jpg')
```

```python
cols = ['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age']
for col in cols:
    Q1 = X[col].quantile(0.25)
    Q3 = X[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    mask = (X[col] >= lower_bound) & (X[col] <= upper_bound)
```

```python
# Filter dataset to remove outliers
X_outlier_detection = X[mask]
y_outlier_detection = y[mask]
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_outlier_detection)
```

```python
q = X_scaled['Insulin'].quantile(.95)
mask = X_scaled['Insulin'] < q
dataNew = X_scaled[mask]
y_outlier_detection = y_outlier_detection[mask]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(dataNew, y_outlier_detection, test_size=0.33, random_state=42)
```

```python
X_train.shape,X_test.shape
```

```python
print(y_train.value_counts())
```

```python
from imblearn.over_sampling import SMOTE
smote = SMOTE(random_state=42)
X_train_resampled, y_train_resampled = smote.fit_resample(X_train, y_train)

# Check resampled class distribution
print("\nResampled class distribution:")
print(pd.Series(y_train_resampled).value_counts())
```

```python
from sklearn.linear_model import LogisticRegression
# Initialize Logistic Regression
classification = LogisticRegression()
# Fit the model to the training data
classification.fit(X_train_resampled, y_train_resampled)
```

```python
y_predictions = classification.predict(X_test)
print(y_predictions)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test, y_predictions)

from sklearn.metrics import classification_report
target_names = ['Non-Diabetic', 'Diabetic']
print(classification_report(y_test, y_predictions, target_names=target_names))
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_predictions)
print(cm)
```

```python
# Get the probability of class 1 (Diabetic)
y_probabilities = classification.predict_proba(X_test)[:, 1]

# Use a lower threshold of 0.4
y_predictions_04 = (y_probabilities >= 0.4).astype(int)

# Evaluate the new predictions
from sklearn.metrics import confusion_matrix, classification_report

print(confusion_matrix(y_test, y_predictions_04))

print(classification_report(
    y_test,
    y_predictions_04,
    target_names=['Non-Diabetic', 'Diabetic']
))
```
